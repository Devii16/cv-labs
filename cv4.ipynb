```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img=cv2.imread('/content/image 3.jpg',cv2.IMREAD_GRAYSCALE)
```

```python
if img is None:
  print("Error:image not found")
else:
  plt.figure(figsize=(12,10))
```

```python
plt.subplot(3,3,1)
plt.imshow(img,cmap='gray')
plt.title("original image")
plt.axis('off')
```

```python
for i in range (8):
  bit_plane = (img >> i) & 1
  vis_plane = bit_plane*255
  plt.subplot(3,3,i+2)
  plt.imshow(bit_plane,cmap='gray')
  plt.title("bit plane {i}")
  plt.axis('off')
```
